1\. Create a random list of number and then save it to a text file named "simple_data.txt"

```python
import numpy as np
import os
rn = np.random.rand(100)

nomefold = "ex5folder"
dirname = os.path.join(os.path.abspath(os.getcwd()),nomefold)
if not os.path.exists(dirname):
    os.mkdir(dirname)
filename = os.path.join(dirname,"simple_data.txt")
np.savetxt(filename,rn)
print(np.loadtxt(filename))
```

2\. Create a random matrix of 5x5 and then save it to a text file named "data.txt"

```python
r5x5 = np.random.rand(5,5)

nomefold = "ex5folder"
dirname = os.path.join(os.path.abspath(os.getcwd()),nomefold)
if not os.path.exists(dirname):
    os.mkdir(dirname)
filename = os.path.join(dirname,"data.txt")
np.savetxt(filename,r5x5)

print(np.loadtxt(filename))
```

3\. Load the saved txt file of point 2 and convert it to a csv file (by hand)

```python
import numpy as np
import csv
import os

# Carica il file txt
load = np.loadtxt(filename)

# Nuovo file CSV
newfilename = os.path.join(dirname, "file_transformed.csv")

# Scrivi il CSV correttamente
with open(newfilename, "w", newline='') as f:
    writer = csv.writer(f, delimiter=",")
    for row in load:
        writer.writerow(row)

# Ricarica il CSV per verificarlo
print(np.loadtxt(newfilename, delimiter=","))
```
